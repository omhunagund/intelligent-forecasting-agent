```python
from pathlib import Path
import pandas as pd

# Project paths
PROJECT_ROOT = Path.cwd().parent
RAW_DATA_DIR = PROJECT_ROOT / "data" / "raw"

print("Project root:", PROJECT_ROOT)
print("Raw data directory:", RAW_DATA_DIR)

# Load the 9 Olist tables
customers = pd.read_csv(RAW_DATA_DIR / "olist_customers_dataset.csv")
geolocation = pd.read_csv(RAW_DATA_DIR / "olist_geolocation_dataset.csv")
order_items = pd.read_csv(RAW_DATA_DIR / "olist_order_items_dataset.csv")
order_payments = pd.read_csv(RAW_DATA_DIR / "olist_order_payments_dataset.csv")
order_reviews = pd.read_csv(RAW_DATA_DIR / "olist_order_reviews_dataset.csv")
orders = pd.read_csv(RAW_DATA_DIR / "olist_orders_dataset.csv")
products = pd.read_csv(RAW_DATA_DIR / "olist_products_dataset.csv")
sellers = pd.read_csv(RAW_DATA_DIR / "olist_sellers_dataset.csv")
category_translation = pd.read_csv(
    RAW_DATA_DIR / "product_category_name_translation.csv"
)

print("\nAll 9 Olist tables loaded successfully.")
```

```python
# Basic structure of every Olist table
tables = {
    "customers": customers,
    "geolocation": geolocation,
    "order_items": order_items,
    "order_payments": order_payments,
    "order_reviews": order_reviews,
    "orders": orders,
    "products": products,
    "sellers": sellers,
    "category_translation": category_translation,
}

for name, df in tables.items():
    print("=" * 70)
    print(f"TABLE: {name}")
    print(f"Rows: {len(df):,}")
    print(f"Columns: {len(df.columns)}")
    print("Columns:")
    print(list(df.columns))
    print()
```

```python
# Compact table summary
summary = []

for name, df in tables.items():
    summary.append({
        "table": name,
        "rows": len(df),
        "columns": len(df.columns),
        "missing_cells": int(df.isna().sum().sum()),
        "duplicate_rows": int(df.duplicated().sum()),
    })

summary_df = pd.DataFrame(summary)

summary_df
```

```python
# Compact schema overview
for name, df in tables.items():
    print(f"{name:20} | {len(df):>9,} rows | {len(df.columns):>2} cols | {', '.join(df.columns)}")
```

```python
# Column-level data types and missing values

for name, df in tables.items():
    print("\n" + "=" * 80)
    print(f"TABLE: {name}")
    print("=" * 80)

    inspection = pd.DataFrame({
        "dtype": df.dtypes.astype(str),
        "missing": df.isna().sum(),
        "missing_pct": (df.isna().mean() * 100).round(2),
        "unique_values": df.nunique(dropna=True),
    })

    print(inspection.to_string())
```

```python
# Missing-value details for the affected tables only

for name in ["orders", "products", "order_reviews"]:
    df = tables[name]

    print("\n" + "=" * 80)
    print(f"TABLE: {name}")
    print("=" * 80)

    missing = df.isna().sum()
    missing = missing[missing > 0]

    if missing.empty:
        print("No missing values")
    else:
        result = pd.DataFrame({
            "missing": missing,
            "missing_pct": (missing / len(df) * 100).round(2)
        })
        print(result.to_string())
```

```python
# Investigate missing-value patterns without modifying the raw data

# 1. Orders: compare missing lifecycle timestamps with order status
order_missing_delivery = (
    orders[
        orders["order_delivered_carrier_date"].isna()
        | orders["order_delivered_customer_date"].isna()
        | orders["order_approved_at"].isna()
    ]
    .groupby("order_status")
    .size()
    .sort_values(ascending=False)
)

print("Orders with at least one missing lifecycle timestamp by order status:")
print(order_missing_delivery.to_string())


# 2. Products: check whether the 610 missing rows are the same rows
product_profile_columns = [
    "product_category_name",
    "product_name_lenght",
    "product_description_lenght",
    "product_photos_qty",
]

product_missing_profile = products[
    products[product_profile_columns].isna().any(axis=1)
]

print("\nProducts with at least one missing profile field:", len(product_missing_profile))

print("\nMissing counts among those product rows:")
print(
    product_missing_profile[product_profile_columns]
    .isna()
    .sum()
)
```

```python
# Impact of missing product/category information on sold items and revenue

# Products with missing category information
missing_category_products = set(
    products.loc[
        products["product_category_name"].isna(),
        "product_id"
    ]
)

# Order items associated with those products
items_missing_category = order_items[
    order_items["product_id"].isin(missing_category_products)
].copy()

# Temporary business value calculation for inspection only
items_missing_category["item_value"] = (
    items_missing_category["price"]
    + items_missing_category["freight_value"]
)

print("Products with missing category:", len(missing_category_products))
print("Order items using those products:", len(items_missing_category))
print("Unique orders affected:", items_missing_category["order_id"].nunique())
print("Total item value affected: {:.2f}".format(
    items_missing_category["item_value"].sum()
))

# Overall comparison
order_items_value = (
    order_items["price"] + order_items["freight_value"]
).sum()

print("\nOverall order-item value: {:.2f}".format(order_items_value))

if order_items_value > 0:
    print(
        "Share of total item value affected: {:.2f}%".format(
            100 * items_missing_category["item_value"].sum()
            / order_items_value
        )
    )
```

```python
# Core key and relationship validation
print("=== CORE KEY UNIQUENESS ===")

checks = {
    "orders.order_id": orders["order_id"].nunique(),
    "customers.customer_id": customers["customer_id"].nunique(),
    "products.product_id": products["product_id"].nunique(),
    "sellers.seller_id": sellers["seller_id"].nunique(),
    "order_items.order_item_id": order_items["order_item_id"].nunique(),
    "category_translation.product_category_name": (
        category_translation["product_category_name"].nunique()
    ),
}

for key, unique_count in checks.items():
    print(f"{key:<45} {unique_count:,}")


print("\n=== REFERENTIAL INTEGRITY ===")

missing_order_refs = (~order_items["order_id"].isin(orders["order_id"])).sum()
missing_product_refs = (~order_items["product_id"].isin(products["product_id"])).sum()
missing_seller_refs = (~order_items["seller_id"].isin(sellers["seller_id"])).sum()
missing_customer_refs = (~orders["customer_id"].isin(customers["customer_id"])).sum()

print(f"Order items with missing order_id reference:     {missing_order_refs:,}")
print(f"Order items with missing product_id reference:   {missing_product_refs:,}")
print(f"Order items with missing seller_id reference:    {missing_seller_refs:,}")
print(f"Orders with missing customer_id reference:       {missing_customer_refs:,}")


print("\n=== ORDER ITEM CARDINALITY ===")

items_per_order = order_items.groupby("order_id").size()

print(f"Orders represented in order_items: {items_per_order.index.nunique():,}")
print(f"Maximum items in a single order:    {items_per_order.max():,}")
print(f"Average items per order:            {items_per_order.mean():.2f}")
print(f"Orders with >1 item:                {(items_per_order > 1).sum():,}")
```

```python
# Validate order-item uniqueness and investigate orders without items

print("=== ORDER ITEM KEY VALIDATION ===")

composite_duplicates = order_items.duplicated(
    subset=["order_id", "order_item_id"]
).sum()

print(
    "Duplicate (order_id, order_item_id) combinations:",
    composite_duplicates
)

print(
    "Unique (order_id, order_item_id) combinations:",
    order_items[["order_id", "order_item_id"]]
    .drop_duplicates()
    .shape[0]
)

print("\n=== ORDERS WITHOUT ORDER ITEMS ===")

orders_without_items = orders[
    ~orders["order_id"].isin(order_items["order_id"])
].copy()

print("Orders without order items:", len(orders_without_items))

print("\nOrder status distribution:")
print(
    orders_without_items["order_status"]
    .value_counts()
    .sort_values(ascending=False)
    .to_string()
)
```

```python
# Validate product-category translation coverage

product_categories = set(
    products["product_category_name"].dropna().unique()
)

translated_categories = set(
    category_translation["product_category_name"].dropna().unique()
)

untranslated_categories = sorted(
    product_categories - translated_categories
)

print("Unique non-null product categories:", len(product_categories))
print("Categories present in translation table:", len(translated_categories))
print("Untranslated product categories:", len(untranslated_categories))

if untranslated_categories:
    print("\nUntranslated categories:")
    for category in untranslated_categories:
        print(category)
else:
    print("\nAll non-null product categories have an English translation.")
```

```python
# Impact of the two untranslated categories on sold items and revenue

untranslated = [
    "pc_gamer",
    "portateis_cozinha_e_preparadores_de_alimentos",
]

untranslated_products = products[
    products["product_category_name"].isin(untranslated)
][["product_id", "product_category_name"]]

untranslated_items = order_items.merge(
    untranslated_products,
    on="product_id",
    how="inner",
)

untranslated_items["item_value"] = (
    untranslated_items["price"]
    + untranslated_items["freight_value"]
)

print("=== UNTRANSLATED CATEGORY IMPACT ===")

print("Products in untranslated categories:",
      untranslated_products["product_id"].nunique())

print("Order items:",
      len(untranslated_items))

print("Unique orders:",
      untranslated_items["order_id"].nunique())

print("Total item value: {:.2f}".format(
    untranslated_items["item_value"].sum()
))

total_value = (
    order_items["price"] + order_items["freight_value"]
).sum()

print("Overall item value: {:.2f}".format(total_value))

print("Share of total item value: {:.2f}%".format(
    100 * untranslated_items["item_value"].sum() / total_value
))

print("\nBreakdown by raw category:")
print(
    untranslated_items.groupby("product_category_name")
    .agg(
        order_items=("order_id", "size"),
        unique_orders=("order_id", "nunique"),
        revenue=("item_value", "sum"),
    )
    .sort_values("revenue", ascending=False)
    .to_string()
)
```

```python
# Inspect the order timeline and purchase-date coverage

purchase_dates = pd.to_datetime(
    orders["order_purchase_timestamp"],
    errors="coerce"
)

print("=== PURCHASE TIMELINE ===")
print("Missing purchase timestamps:", purchase_dates.isna().sum())
print("Earliest purchase:", purchase_dates.min())
print("Latest purchase:", purchase_dates.max())
print("Number of unique purchase dates:", purchase_dates.dt.date.nunique())

print("\n=== ORDERS BY STATUS ===")
print(
    orders["order_status"]
    .value_counts()
    .sort_values(ascending=False)
    .to_string()
)

print("\n=== ORDERS BY PURCHASE MONTH ===")
monthly_orders = (
    purchase_dates
    .dt.to_period("M")
    .value_counts()
    .sort_index()
)

print(monthly_orders.to_string())
```

```python
# Temporary EDA revenue series
# No raw data is modified.

# Copy only the fields needed for this inspection
revenue_items = order_items[
    ["order_id", "price", "freight_value"]
].copy()

revenue_items["item_revenue"] = (
    revenue_items["price"] + revenue_items["freight_value"]
)

# Attach purchase timestamp
revenue_eda = revenue_items.merge(
    orders[["order_id", "order_purchase_timestamp"]],
    on="order_id",
    how="inner",
    validate="many_to_one",
)

revenue_eda["purchase_date"] = pd.to_datetime(
    revenue_eda["order_purchase_timestamp"]
).dt.date

# Daily revenue
daily_revenue = (
    revenue_eda
    .groupby("purchase_date")["item_revenue"]
    .sum()
    .sort_index()
)

print("=== DAILY REVENUE COVERAGE ===")
print("Earliest revenue date:", daily_revenue.index.min())
print("Latest revenue date:", daily_revenue.index.max())
print("Number of days with revenue:", len(daily_revenue))

print("\n=== TOTAL REVENUE ===")
print(f"Total revenue: {daily_revenue.sum():,.2f}")

print("\n=== FIRST 15 DAYS ===")
print(daily_revenue.head(15).to_string())

print("\n=== LAST 15 DAYS ===")
print(daily_revenue.tail(15).to_string())
```

```python
# Temporary weekly revenue series for EDA

daily_revenue_series = pd.Series(
    daily_revenue.values,
    index=pd.to_datetime(daily_revenue.index)
).sort_index()

weekly_revenue = (
    daily_revenue_series
    .resample("W-SUN")
    .sum()
)

print("=== WEEKLY REVENUE COVERAGE ===")
print("First week:", weekly_revenue.index.min())
print("Last week:", weekly_revenue.index.max())
print("Number of weekly periods:", len(weekly_revenue))

print("\n=== ZERO-REVENUE WEEKS ===")
print("Number of zero-revenue weeks:", (weekly_revenue == 0).sum())

print("\n=== FIRST 15 WEEKS ===")
print(weekly_revenue.head(15).to_string())

print("\n=== LAST 15 WEEKS ===")
print(weekly_revenue.tail(15).to_string())
```

```python
# Revenue concentration by month

monthly_revenue = (
    daily_revenue_series
    .resample("MS")
    .sum()
)

print("=== MONTHLY REVENUE ===")
print(monthly_revenue.to_string())

print("\n=== MONTHS WITH VERY LOW REVENUE ===")
print(
    monthly_revenue[monthly_revenue < monthly_revenue.max() * 0.05]
    .to_string()
)
```

```python
# Check weekly continuity inside the approved modeling window

MODEL_START = pd.Timestamp("2017-01-01")
MODEL_END = pd.Timestamp("2018-08-26")

weekly_model = weekly_revenue.loc[
    (weekly_revenue.index >= MODEL_START)
    & (weekly_revenue.index <= MODEL_END)
].copy()

# Create the complete expected weekly calendar
expected_weeks = pd.date_range(
    start=weekly_model.index.min(),
    end=weekly_model.index.max(),
    freq="W-SUN"
)

missing_calendar_weeks = expected_weeks.difference(weekly_model.index)

print("=== APPROVED MODELING WINDOW ===")
print("Start:", weekly_model.index.min())
print("End:", weekly_model.index.max())
print("Observed weekly periods:", len(weekly_model))
print("Expected weekly periods:", len(expected_weeks))

print("\n=== CALENDAR CONTINUITY ===")
print("Missing calendar weeks:", len(missing_calendar_weeks))

if len(missing_calendar_weeks) > 0:
    print("\nMissing weeks:")
    print(missing_calendar_weeks.to_list())

print("\n=== ZERO-REVENUE WEEKS WITHIN MODEL WINDOW ===")
zero_weeks = weekly_model[weekly_model == 0]
print("Count:", len(zero_weeks))

if len(zero_weeks) > 0:
    print(zero_weeks.to_string())
```

```python
# Impact of the approved order-status revenue rule

EXCLUDED_STATUSES = {"canceled", "unavailable"}

# Build the temporary revenue view directly from the loaded raw tables
revenue_status_eda = order_items[
    ["order_id", "price", "freight_value"]
].copy()

revenue_status_eda["item_revenue"] = (
    revenue_status_eda["price"] + revenue_status_eda["freight_value"]
)

# Attach order status
revenue_status_eda = revenue_status_eda.merge(
    orders[["order_id", "order_status"]],
    on="order_id",
    how="inner",
    validate="many_to_one",
)

# Revenue by status
status_summary = (
    revenue_status_eda
    .groupby("order_status")
    .agg(
        order_items=("order_id", "size"),
        unique_orders=("order_id", "nunique"),
        revenue=("item_revenue", "sum"),
    )
    .sort_values("revenue", ascending=False)
)

print("=== REVENUE BY ORDER STATUS ===")
print(status_summary.to_string())

# Excluded vs included revenue
excluded = revenue_status_eda[
    revenue_status_eda["order_status"].isin(EXCLUDED_STATUSES)
]

included = revenue_status_eda[
    ~revenue_status_eda["order_status"].isin(EXCLUDED_STATUSES)
]

excluded_revenue = excluded["item_revenue"].sum()
included_revenue = included["item_revenue"].sum()
total_revenue = revenue_status_eda["item_revenue"].sum()

print("\n=== APPROVED REVENUE RULE IMPACT ===")
print(f"Total item revenue before status filter: {total_revenue:,.2f}")
print(f"Excluded revenue: {excluded_revenue:,.2f}")
print(f"Included revenue: {included_revenue:,.2f}")

print(
    f"Excluded revenue share: "
    f"{100 * excluded_revenue / total_revenue:.2f}%"
)

print(
    f"Excluded order-item rows: {len(excluded):,} "
    f"of {len(revenue_status_eda):,} "
    f"({100 * len(excluded) / len(revenue_status_eda):.2f}%)"
)

print(f"Excluded unique orders: {excluded['order_id'].nunique():,}")
```

```python
# Final EDA revenue series using the approved business definition

EXCLUDED_STATUSES = {"canceled", "unavailable"}

# Start from order items
revenue_final_eda = order_items[
    ["order_id", "product_id", "seller_id", "price", "freight_value"]
].copy()

# Calculate item revenue
revenue_final_eda["item_revenue"] = (
    revenue_final_eda["price"] + revenue_final_eda["freight_value"]
)

# Attach purchase timestamp and order status
revenue_final_eda = revenue_final_eda.merge(
    orders[
        [
            "order_id",
            "customer_id",
            "order_purchase_timestamp",
            "order_status",
        ]
    ],
    on="order_id",
    how="inner",
    validate="many_to_one",
)

# Apply approved status rule
revenue_final_eda = revenue_final_eda[
    ~revenue_final_eda["order_status"].isin(EXCLUDED_STATUSES)
].copy()

# Convert purchase timestamp for time-series aggregation
revenue_final_eda["purchase_timestamp"] = pd.to_datetime(
    revenue_final_eda["order_purchase_timestamp"]
)

# Daily revenue
daily_revenue_final = (
    revenue_final_eda
    .set_index("purchase_timestamp")["item_revenue"]
    .resample("D")
    .sum()
)

# Weekly revenue
weekly_revenue_final = (
    daily_revenue_final
    .resample("W-SUN")
    .sum()
)

print("=== FINAL EDA REVENUE SERIES ===")
print(f"Item rows: {len(revenue_final_eda):,}")
print(f"Total revenue: {revenue_final_eda['item_revenue'].sum():,.2f}")

print("\nDaily series:")
print(f"Start: {daily_revenue_final.index.min()}")
print(f"End:   {daily_revenue_final.index.max()}")

print("\nWeekly series:")
print(f"Start: {weekly_revenue_final.index.min()}")
print(f"End:   {weekly_revenue_final.index.max()}")
print(f"Weekly periods: {len(weekly_revenue_final):,}")
```

```python
# Approved modeling window

MODEL_START = pd.Timestamp("2017-01-01")
MODEL_END = pd.Timestamp("2018-08-26")

weekly_revenue_model = weekly_revenue_final.loc[
    (weekly_revenue_final.index >= MODEL_START)
    & (weekly_revenue_final.index <= MODEL_END)
].copy()

print("=== APPROVED MODELING SERIES ===")
print(f"Start: {weekly_revenue_model.index.min()}")
print(f"End:   {weekly_revenue_model.index.max()}")
print(f"Periods: {len(weekly_revenue_model):,}")
print(f"Total revenue in modeling window: {weekly_revenue_model.sum():,.2f}")
print(f"Zero-revenue weeks: {(weekly_revenue_model == 0).sum()}")
```

```python
# Region analysis using the approved 5-region structure

STATE_TO_REGION = {
    # North
    "AC": "North",
    "AP": "North",
    "AM": "North",
    "PA": "North",
    "RO": "North",
    "RR": "North",
    "TO": "North",

    # Northeast
    "AL": "Northeast",
    "BA": "Northeast",
    "CE": "Northeast",
    "MA": "Northeast",
    "PB": "Northeast",
    "PE": "Northeast",
    "PI": "Northeast",
    "RN": "Northeast",
    "SE": "Northeast",

    # Central-West
    "DF": "Central-West",
    "GO": "Central-West",
    "MT": "Central-West",
    "MS": "Central-West",

    # Southeast
    "ES": "Southeast",
    "MG": "Southeast",
    "RJ": "Southeast",
    "SP": "Southeast",

    # South
    "PR": "South",
    "SC": "South",
    "RS": "South",
}

# Attach customer state and region
region_revenue_eda = revenue_final_eda.merge(
    customers[["customer_id", "customer_state"]],
    on="customer_id",
    how="left",
    validate="many_to_one",
)

region_revenue_eda["region"] = (
    region_revenue_eda["customer_state"].map(STATE_TO_REGION)
)

print("=== REGION MAPPING VALIDATION ===")
print("Missing state values:", region_revenue_eda["customer_state"].isna().sum())
print("Missing region mappings:", region_revenue_eda["region"].isna().sum())

print("\n=== REVENUE BY MACRO-REGION ===")

region_summary = (
    region_revenue_eda
    .groupby("region")
    .agg(
        order_items=("order_id", "size"),
        unique_orders=("order_id", "nunique"),
        revenue=("item_revenue", "sum"),
    )
    .sort_values("revenue", ascending=False)
)

region_summary["revenue_share_pct"] = (
    region_summary["revenue"]
    / region_summary["revenue"].sum()
    * 100
).round(2)

print(region_summary.to_string())
```

```python
# Reconcile regional revenue against the approved total revenue

base_total = revenue_final_eda["item_revenue"].sum()
region_total = region_revenue_eda["item_revenue"].sum()

print("=== REVENUE RECONCILIATION ===")
print(f"Approved revenue total: {base_total:,.2f}")
print(f"Regional revenue total: {region_total:,.2f}")
print(f"Difference: {region_total - base_total:,.2f}")

print("\n=== ROW COUNT RECONCILIATION ===")
print(f"Revenue EDA rows: {len(revenue_final_eda):,}")
print(f"Regional EDA rows: {len(region_revenue_eda):,}")

print("\n=== CUSTOMER JOIN DUPLICATION CHECK ===")
customer_counts = customers["customer_id"].value_counts()
print("Customer IDs occurring more than once:", (customer_counts > 1).sum())

print("\n=== REGION VALUE BY REGION ===")
print(
    region_revenue_eda.groupby("region")["item_revenue"]
    .sum()
    .sort_values(ascending=False)
    .to_string()
)
```

```python
# Raw product-category analysis
# Uses known categories only, as previously approved.
# Overall revenue is NOT affected by this filtering.

# Add product category to revenue-bearing order items
category_revenue_eda = revenue_final_eda.merge(
    products[["product_id", "product_category_name"]],
    on="product_id",
    how="left",
    validate="many_to_one",
)

# Exclude only genuinely missing categories from category-level analysis
category_revenue_eda = category_revenue_eda[
    category_revenue_eda["product_category_name"].notna()
].copy()

# Manually translate the two categories not covered by the translation table
MANUAL_TRANSLATIONS = {
    "pc_gamer": "PC Gaming",
    "portateis_cozinha_e_preparadores_de_alimentos":
        "Kitchen Appliances / Food Preparation Appliances",
}

category_revenue_eda = category_revenue_eda.merge(
    category_translation,
    on="product_category_name",
    how="left",
    validate="many_to_one",
)

category_revenue_eda["product_category_name_english"] = (
    category_revenue_eda["product_category_name_english"]
    .fillna(category_revenue_eda["product_category_name"].map(MANUAL_TRANSLATIONS))
)

# Aggregate
category_summary = (
    category_revenue_eda
    .groupby(
        "product_category_name_english",
        dropna=False
    )
    .agg(
        order_items=("order_id", "size"),
        unique_orders=("order_id", "nunique"),
        revenue=("item_revenue", "sum"),
    )
    .sort_values("revenue", ascending=False)
)

category_summary["revenue_share_pct"] = (
    category_summary["revenue"]
    / category_summary["revenue"].sum()
    * 100
).round(2)

print("=== RAW CATEGORY DISTRIBUTION ===")
print(f"Categories represented in category analysis: {len(category_summary)}")
print(
    f"Revenue covered by category analysis: "
    f"{category_summary['revenue'].sum():,.2f}"
)

print("\n=== ALL RAW CATEGORIES BY REVENUE ===")
print(category_summary.to_string())
```

```python
# Compact category distribution for grouping decisions

category_view = category_summary.reset_index().copy()

category_view["cumulative_revenue_share_pct"] = (
    category_view["revenue_share_pct"].cumsum().round(2)
)

print("=== TOP 25 RAW CATEGORIES ===")
print(
    category_view[
        [
            "product_category_name_english",
            "order_items",
            "unique_orders",
            "revenue",
            "revenue_share_pct",
            "cumulative_revenue_share_pct",
        ]
    ]
    .head(25)
    .to_string(index=False)
)

print("\n=== BOTTOM 25 RAW CATEGORIES ===")
print(
    category_view[
        [
            "product_category_name_english",
            "order_items",
            "unique_orders",
            "revenue",
            "revenue_share_pct",
        ]
    ]
    .tail(25)
    .to_string(index=False)
)
```

```python
# Category concentration summary

print("=== CATEGORY CONCENTRATION ===")

for n in [5, 10, 15, 20, 25, 30]:
    share = category_view.head(n)["revenue_share_pct"].sum()
    print(f"Top {n:>2} categories revenue share: {share:.2f}%")

print("\nCategories by revenue-share threshold:")

for threshold in [1.0, 0.5, 0.1, 0.05]:
    count = (category_view["revenue_share_pct"] >= threshold).sum()
    print(
        f"Categories contributing >= {threshold:.2f}% each: {count}"
    )
```

```python
# ============================================================
# CLEAN CATEGORY ANALYSIS STATE
# Prepare the category-level EDA dataframe.
#
# Revenue definition is already finalized in revenue_final_eda.
# This cell only attaches product categories and translations.
# Raw source data remains unchanged.
# ============================================================

# 1. Attach the raw product category to the approved revenue dataset
clean_category = revenue_final_eda.merge(
    products[
        ["product_id", "product_category_name"]
    ],
    on="product_id",
    how="left",
    validate="many_to_one",
)

# 2. Exclude genuinely missing product categories
#    from category-level analysis only.
clean_category = clean_category[
    clean_category["product_category_name"].notna()
].copy()

# 3. Attach the Olist English category translation
clean_category = clean_category.merge(
    category_translation[
        [
            "product_category_name",
            "product_category_name_english",
        ]
    ],
    on="product_category_name",
    how="left",
    validate="many_to_one",
)

# 4. Approved manual translations for the two categories
#    missing from the official translation table.
MANUAL_TRANSLATIONS = {
    "pc_gamer": "PC Gaming",
    "portateis_cozinha_e_preparadores_de_alimentos":
        "Kitchen Appliances / Food Preparation Appliances",
}

# 5. Create the canonical category column.
#    Alias normalization is deliberately handled in the
#    separate CATEGORY_ALIASES cell that follows.
clean_category["canonical_category"] = (
    clean_category["product_category_name_english"]
    .fillna(
        clean_category["product_category_name"].map(
            MANUAL_TRANSLATIONS
        )
    )
)

# 6. Basic validation
print("=== CLEAN CATEGORY ANALYSIS STATE ===")

print(
    "Rows in category analysis:",
    f"{len(clean_category):,}"
)

print(
    "Unique raw categories:",
    clean_category["product_category_name"].nunique()
)

print(
    "Unique canonical categories before alias normalization:",
    clean_category["canonical_category"].nunique()
)

print(
    "Untranslated categories:",
    clean_category["canonical_category"].isna().sum()
)

if clean_category["canonical_category"].isna().any():
    print("\nUntranslated raw categories:")
    print(
        clean_category.loc[
            clean_category["canonical_category"].isna(),
            "product_category_name"
        ].drop_duplicates().to_list()
    )
else:
    print("✓ All category values have a canonical label.")
```
